```python
import os
import xarray as xr

from xcube.core.store import new_data_store
```

```python
S3_USER_STORAGE_KEY = os.environ["S3_USER_STORAGE_KEY"]
S3_USER_STORAGE_SECRET = os.environ["S3_USER_STORAGE_SECRET"]
S3_USER_STORAGE_BUCKET = os.environ["S3_USER_STORAGE_BUCKET"]
```

```python
team_store = new_data_store(
    "s3",
    root=S3_USER_STORAGE_BUCKET,
    storage_options=dict(
        anon=False, key=S3_USER_STORAGE_KEY, secret=S3_USER_STORAGE_SECRET
    ),
)
```

```python
list(team_store.get_data_ids())
```

```python
input_store = new_data_store(
    "file",
    root="../workflow/out/"
)
```

```python
input_store.list_data_ids()
```

```python
ds = input_store.open_data("sss_anomalies.zarr")
ds
```

```python
ds.attrs["title"] = "Sea Surface Salinity Monthly Anomalies 2010 - 2023"
ds
```

```python
ds.sss.attrs.pop("ancilliary")
ds.sss.attrs["long_name"] = "Sea Surface Salinity Anomaly"
ds.sss.attrs["standard_name"] = "sea_surface_salinity_anomaly"
ds.sss.attrs["valid_min"] = -19
ds.sss.attrs["valid_max"] = 19
ds = ds.rename(sss="sss_anom")
```

```python
ds
```

```python
team_store.write_data(
    ds,
    "sss_anom.zarr",
    replace=True
)
```
